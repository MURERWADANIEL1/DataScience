```python
import numpy as np
import matplotlib.pyplot as plt
from numpy.fft import fft2, ifft2, fftshift, ifftshift
from PIL import Image

# Load local image directly (no urllib needed)
file_path = r'C:\Users\MURERWA DANIEL\Downloads\120px-Example.png'
img = Image.open(file_path).convert('L')  # Convert to grayscale
img_array = np.array(img)

# Simulate k-space (Fourier transform)
k_space = fftshift(fft2(img_array))

# Simulate inverse Fourier transform (reconstruction)
reconstructed = np.abs(ifft2(ifftshift(k_space)))

# Display results
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.title("Original Image")
plt.imshow(img_array, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Simulated k-space (log scale)")
plt.imshow(np.log(1 + np.abs(k_space)), cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Reconstructed Image")
plt.imshow(reconstructed, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```
