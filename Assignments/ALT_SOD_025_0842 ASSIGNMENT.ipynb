##TASK
Analyze the fashion store dataset using filtering and subsetting to extract the product name, color, and brand name for products with a cost price between 10.00 and 30.00.

```python
import pandas as pd

my_data=r'c:\Users\MURERWA DANIEL\Downloads\dataset_fashion_store_products.csv'
df=pd.read_csv(my_data)
df.head()
```

```python
#filtering the data for the proce range or 10 and 30
df_filtered=df[(df['cost_price']>=10.00) & (df['cost_price']<=30.00)]
df_filtered[['product_name', 'color','brand']]
```

```python
#subsetting the data to get the product name, color, and brand name
df_subset = df_filtered[['product_name', 'color', 'brand']]
df_subset
```
