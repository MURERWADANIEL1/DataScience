```python
#imports
import mailbox
import pandas as pd
import re
from email.header import decode_header
```

```python
def decode_mime_words(s):
    decoded = ''
    for word, charset in decode_header(s):
        if isinstance(word, bytes):
            # If encoding is unknown or invalid, fall back to utf-8
            try:
                decoded += word.decode(charset or 'utf-8', errors='replace')
            except LookupError:
                decoded += word.decode('utf-8', errors='replace')
        else:
            decoded += word
    return decoded
```

```python
#load the mailbox
mbox = mailbox.mbox(r"C:\Users\MURERWA DANIEL\Downloads\All mail Including Spam and Trash-001.mbox")
```

```python
#exracting data from the emails
data = []
for message in mbox:
    subject = decode_mime_words(message['subject']) if message['subject'] else None
    sender = decode_mime_words(message['from']) if message['from'] else None
    recipient = decode_mime_words(message['to']) if message['to'] else None
    date = message['date']
    label = message['X-Gmail-Labels'] if 'X-Gmail-Labels' in message else None
    thread_id = message['X-GM-THRID'] if 'X-GM-THRID' in message else None
    
    data.append([subject, sender, date, recipient, label, thread_id])
```

```python
#create the dataframe
gmail_data = pd.DataFrame(data, columns=["subject", "from", "date", "to", "label", "thread"])
gmail_data.head(20)
```
