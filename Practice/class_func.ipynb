```python
def grade_checker():
    try:
        x=int(input("Enter Your Marks:"))
        if x>=80 and x<=100:
            print("Your Grade is A")

        elif x<80 and x>=60:
            print("Your grade is B")

        elif x<60 and x>=40:
            print("Your Grade is C")
        
        elif x>0 and x<40:
            print("Your Grade is D")
            print("You are Failed")

        else:
            print("Invalid Marks Entered")
    except ValueError:
        print("Please enter a valid number for marks.")
grade_checker()
```

```python
class Employee:
    def __init__(self, name, age, salary):
        self.name=name
        self.age=age
        self.salary=salary

    def promote(self, raise_salary):
        self.salary+=raise_salary
        return f'{self.name} has new salary of {self.salary}'
    
    def retire(self):
        return f'{self.name} has retired at age of {self.age} with a salary of {self.salary}'
    
    def main():
        emp1=Employee("John", 30, 50000)
        print(emp1.promote(5000))
        print(emp1.retire())

if __name__=="__main__":
    Employee.main()
```

```python
class ComplexMaths():
    def __init__(self, real, imaginary):
        self.real=real
        self.imaginary=imaginary

    def __str__(self):

        if self.imaginary>=0:
            return f"{self.real}+{self.imaginary}i"
        else:
            return f"{self.real}-{abs(self.imaginary)}i"
     
    def addition(self, other_complex):

        if not isinstance(other_complex, ComplexMaths):
            raise TypeError("Operand should be a complex Number")
        new_real=self.real+other_complex.real
        new_imaginary=self.imaginary+other_complex.imaginary

        return ComplexMaths(new_real, new_imaginary)
    
    def subtraction(self, other_complex):
        if not isinstance(other_complex, ComplexMaths):
            raise TypeError("Operand must be an instance of Complex Number")
        
        new_real=self.real-other_complex.real
        new_imaginary=self.imaginary-other_complex.imaginary
        return ComplexMaths(new_real, new_imaginary)
    
    def main():
        num1=ComplexMaths(5, 9)
        num2=ComplexMaths(2,3)

        print(f'Number 1: {num1}')
        print(f'Number 2: {num2}')

        sum_result=num1.addition(num2)
        print(f'Addition ({num1}+{num2}): {sum_result}')

        diff_result=num1.subtraction(num2)
        print(f'Subtraction ({num1}-{num2}): {diff_result}')

        diff_result_2=num2.subtraction(num1)
        print(f'Subtraction ({num2}-{num2}): {diff_result_2}')

if __name__=="__main__":
    ComplexMaths.main()   
```

```python
class AgeTest():

    def __init__(self, name, age):
        self.name=name
        if not isinstance(age, (int, float)):

            raise ValueError("Age must be an integer")
        if age<0:
            raise ValueError("Age can't be negative")
        self.age=age
    

    def check_age_requirement(self, required_age):
        if not isinstance(required_age, (int, float)):
            return f"Invalid input for {self.name}: "
        if required_age<0:
            return f"Invalid input for {self.name}: Required age cannot be negative."

        if self.age >= required_age:
            return f"{self.name} (age {self.age}) meets or exceeds the requirement of {required_age} years."
        else:
            return f"{self.name} (age {self.age}) does not meet the requirement of {required_age} years."

    @staticmethod
    def main():
        try:
            person1 = AgeTest("Kimathi", 76)
            person2 = AgeTest("Aisha", 17)
            person3 = AgeTest("Zane", 30)

            print(person1.check_age_requirement(65)) 
            print(person2.check_age_requirement(18)) 
            print(person3.check_age_requirement(30)) 
            print(person3.check_age_requirement(35))  
            print(person1.check_age_requirement("twenty"))
        except ValueError as e:
            print(f"Error during instantiation: {e}")

if __name__ == "__main__":
    AgeTest.main()
```
