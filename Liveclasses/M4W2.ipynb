Filterin
    process of subsetting rows based on boolean conditions
    it allows to focus on relevant data for analysis and visualization

Use cases
    HR and recruitment Analytics
    Filter employees or job seekers based on some conditions
    Business value: employee management and talent recruitment

    Market Analytics:
        filtering for target customers based in buying psychology/demographics
        
    Business Value:
        Profit/sales/high conversion targets

    Retail Analytics:
        Inventory of product status
        business value-rordering model

    Financial Analytics:
        audit transcations for certain thresholds
        business value-fraud detection and financial compliance

    Health Analytics
        Tracking Patient visits
        business value-build model for reminders 

```python
import pandas as pd # Data manipulation library
```

```python
#create a mocking dataframe
student_data={
    'student_name':['John Mark', 'Precious Jane', 'Mark James Doe', 'olivia Grace', 'Daniel Murerwa'],
    'city':['Lagos', 'Kaduma', 'Lome', 'Kigali', 'Nairobi'],
    'age':[23, 21, 25, 20, 22],
    'scores':[85, 90, 78, 88, 92],
    }
```

```python
df=pd.DataFrame(student_data)
df.head() # Display the first few rows of the DataFrame
```

```python
#filteing rows for scores
df['scores']>85
```

```python
df['age']>25
```

```python
df[df['scores']>=90]
```

```python
df_90=df[df['scores']>=90]
df['city']=='Kigali'
```

```python
#filter for city and scores
#city is lagos and scores is greater than  70
df['city']=='Lagos' 
df['scores']>70
```
