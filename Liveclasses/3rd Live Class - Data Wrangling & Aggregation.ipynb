##### *Technical Definition of Data Wrangling*

Data wrangling, also known as data munging or data preparation, is the process of cleaning, structuring, and transforming raw data into a usable format for analysis and other data-driven applications. It addresses issues like missing values, inconsistencies, and duplicates to ensure data quality and make it more consumable

Data wrangling, sometimes referred to as data munging, is the process of transforming and mapping data from one "raw" data form into another format with the intent of making it more appropriate and valuable for a variety of downstream purposes such as analytics. The goal of data wrangling is to assure quality and useful data. Data analysts typically spend the majority of their time in the process of data wrangling compared to the actual analysis of the data.

##### *A simple example of a groupby function*

```python
# Create a simple dataframe
import pandas as pd
```

```python
data = {
    'Store Location': ['Lagos', 'Lagos', 'Abuja', 'Abuja', 'Kano'],
    'Product': ['Bread', 'Milk', 'Bread', 'Milk', 'Bread'],
    'Sales': [100, 200, 150, 250, 50]
}
```

```python
df = pd.DataFrame(data)
df
```

##### *For a store location, what is their total sales?*

```python
?df.groupby
```

```python
df.groupby('Store Location')['Sales']
```

```python
df.groupby('Store Location')['Sales'].sum()
```

##### *Other Aggregation Functions*

.sum()

.mean()

.median()

.count()

.max()/min()

.agg()

```python
df.groupby('Store Location')['Sales'].agg(['sum', 'mean', 'count'])
```

#### Use groupby on a dataset

```python
housing_data = pd.read_csv('house_data.csv')
house_data.head()
```

##### *Are house prices increasing or decreasing by year?*

```python
# Year Sold
# House Price
```

```python
house_data.groupby('YrSold')['SalePrice'].mean()
```

```python
house_data.groupby('YrSold')['SalePrice'].median()
```

##### *Do land size increase based on housing zones?*

```python
house_data['MSZoning'].head()
```

```python
# Feature Transformation 
house_data['MSZoning'].unique()
```

```python
zoning_housing = {"RL": "Residential Low Density",
"RM": "Residential Medium Density",
"C (all)": "Commercial",
"FV": "Floating Village Residential",
"RH": "Floating Village Residential"}
```

```python
house_data['Zone Identifier'] = house_data['MSZoning'].replace(zoning_housing)
```

```python
house_data.head()
```

```python
house_data['Zone Identifier'].value_counts(normalize=True)
```

```python
house_data['LotArea'].head()
```

```python
# Groupby function
house_data.groupby('Zone Identifier')['LotArea'].median()
```

```python
house_data.groupby('Zone Identifier')['LotArea'].agg(['median', 'mean', 'sum'])
```

##### *Sales based on zones*

```python
house_data.groupby('Zone Identifier')['SalePrice'].agg(['mean', 'median', 'count'])
```

##### *Average sale price based on year sold & zone*
