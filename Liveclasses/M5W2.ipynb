```python
#what does BHK mean?
#BHK means Bedroom, Hall, Kitchen

#Feature engineer the size column and create a new column called bhk

df_2['bhk']=df_2['size'].apply(lambda x: int(x.split(" ")[0]))
df_2.head(10)
```

**New Pipeline**

```python
df_3=df_2.copy()
```

```python
#evaluate total_sqft column
df_3['total_sqft'].unique()
```

```python
#convert all the values in the total_sqft column to float
#if it's not a float value, let it be

def convert_float(x):
    try:
        float(x)
    except:
        return False
    
    return True
```

```python
#apply the convert_float function to the total_sqft column
df_3['total_sqft'].apply(convert_float)
```

```python
df_3[df_3['total_sqft'].apply(convert_float)]
```

```python
df_3[~df_3['total_sqft'].apply(convert_float)] #negate
```

```python
def convert_sqft(x):
    tokens=x.split('-')
    if len(tokens)==2:
        return(float(tokens[0])+float(tokens[1]))/2
    
    try:
        return float(x)
    except:
        return None
```

```python
convert_sqft('650 - 655')
```

```python
df_3['total_sqft']=df_3['total_sqft'].apply(convert_sqft)
df_3.head(30)
```

```python
df_3.info()
```

```python
df_4=df_3[df_3['total_sqft'].notnull()]
```

```python
df_4.info()
```
