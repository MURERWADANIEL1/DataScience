Technical definition

Data aggregation is a process of applying summary statistics to grouped subsets of data 
Done using groupby method of Pandas


Non-technical Definition

Groupby similar things together and summarizing them so its easier to understand

Why is data aggregation important in real-world problem solving?

Retail usecase -> Summarize product sales 
Public Policy usecase -> 
Group by demographics
Group by location to understand poverty distributions
Group by location to understand disease breakout
Group by location to undeerstand food  insecurities
Logistic Usecase- groupby by cities to understand delivery 

```python
import pandas as pd
data_frame={
    'Store':['Lagos', 'Abuja', 'Port Harcourt', 'Ibadan', 'Lagos'],
    'Product':['Rice', 'Beans', 'Maize', 'Wheat', 'Rice'],
    'Sales':[1500, 2000, 1800, 2200, 1600],
}
```

```python
df= pd.DataFrame(data_frame)
df
```

```python
#for a store location, what is their total sales?
?df.groupby
```

```python
df.groupby('Store')['Sales']
```

```python
df.groupby('Store')['Sales'].sum()
```

Aggregation functions

.sum()
.mean()
.median()
.count()
max()/min()
agg()

```python
df.groupby('Store')['Sales'].agg('sum', 'mean')
```
