**DATA WRANGLING AND MISSING DATA**

```python
import pandas as pd
import numpy as np
```

```python
#Pandas doesn't use None to rep missing values, it uses np.nan
value=np.array([1, None, 2, 3])
value
```

```python
#value.sum() this fails because None is not a number
#NaN means missing numeric data

value1=np.array([1, np.nan, 2, 3])
value1
```

```python
value1.sum()
```

```python
np.nansum(value1)  # This will ignore NaN values in the sum
```

Types of Missing Value

    1. missing completely at random(MCAR)
     
     intuition-someone skipping survey question for no reason
     Example, someone forget to enter ther details in a survey-internet failure
     this data is missing and has no connection with any other variable


     2. missing at random (MAR)
     intuition-missing value depends on oserved data
     example-employees are less likely to share their earning publicly

     3. Missing not at random (MNAR)

     intuition-missing depending on unobserved data
     example-persons with low income don't disclose it, people don't want to disclose some information (health, employment behavior...)
